## DATA TRANSFORMATIONS

### Checking the missing values

```python
df.isnull().sum()
```

### Filling the missing values

>The nan values werent just filled by the overall median of review rating but instead they were filled by the overall median of their corresponding category.This removes any bias

```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

### Standardizing the Columns

```python
df.columns=df.columns.str.lower().str.replace(" ","_")
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
df.columns
```

### Feature Engineering

#### 1. Creating an age_group column

```python
labels=['Young Adult','Adult','Middle-Aged','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df.head(5)
```

#### 2. Creating a purchase_frequency_days columns

```python
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}
df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['frequency_of_purchases','purchase_frequency_days']]
```
